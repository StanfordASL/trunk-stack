```python
import pandas as pd
import numpy as np

# === Parameters ===
input_path = "/home/trunk/Documents/trunk-stack/stack/main/data/trajectories/dynamic/controls_120.csv"
output_path = "/home/trunk/Documents/trunk-stack/stack/main/data/trajectories/dynamic/control_inputs_controlled_503.csv"

# Define clipping ranges
tip_range, mid_range, base_range = 111, 71, 51

# === Load CSV ===
df = pd.read_csv(input_path, header=None)
df.columns = [f"u{i+1}" for i in range(df.shape[1])]
df.insert(0, "ID", range(len(df)))

# === Apply clipping using JAX ===
u1 = np.clip(df["u1"].to_numpy(), -base_range, base_range)
u2 = np.clip(df["u2"].to_numpy(), -tip_range, tip_range)
u3 = np.clip(df["u3"].to_numpy(), -mid_range, mid_range)  
u4 = np.clip(df["u4"].to_numpy(), -tip_range, tip_range)
u5 = np.clip(df["u5"].to_numpy(), -mid_range, mid_range)
u6 = np.clip(df["u6"].to_numpy(), -base_range, base_range)

# === Replace columns with clipped values ===
df["u1"] = u1
df["u2"] = u2
df["u3"] = u3
df["u4"] = u4
df["u5"] = u5
df["u6"] = u6

# === Save result ===
df.to_csv(output_path, index=False)
print(f"Saved clipped CSV to: {output_path}")
```
